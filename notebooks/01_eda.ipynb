# Customer Churn Prediction - EDA
This notebook explores the IBM Telco Customer Churn CSV. It does not create or invent data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
DATA_PATH = PROJECT_ROOT / 'data' / 'raw' / 'WA_Fn-UseC_-Telco-Customer-Churn.csv'
if not DATA_PATH.exists():
    raise FileNotFoundError(f'Place the IBM Telco Customer Churn CSV at {DATA_PATH}')
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
df.dtypes
```

```python
print(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
df.describe(include='all').T
```

```python
sns.countplot(data=df, x='Churn')
plt.title('Churn Distribution')
plt.show()
```

```python
clean = df.copy()
clean['TotalCharges'] = pd.to_numeric(clean['TotalCharges'], errors='coerce')
sns.boxplot(data=clean, x='Churn', y='MonthlyCharges')
plt.title('Monthly Charges by Churn')
plt.show()
```

```python
sns.countplot(data=df, x='Contract', hue='Churn')
plt.title('Churn by Contract Type')
plt.xticks(rotation=15)
plt.show()
```

## Initial observations
- Check whether churn classes are imbalanced.
- `TotalCharges` may require numeric conversion because blank values can be stored as text.
- Contract type, tenure, and monthly charges are useful variables to inspect for churn patterns.
- The final preprocessing pipeline handles missing values and categorical encoding after the train/test split, preventing leakage.
